```python
# ==========================
# STEP 1: LOAD AND TRAIN MODEL
# ==========================

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Load datasets

train_df = pd.read_csv("/content/data/Training.csv")
test_df = pd.read_csv("/content/data/Testing.csv")

# Remove unnamed column if it exists

train_df = train_df.loc[:, ~train_df.columns.str.contains("^Unnamed")]
test_df = test_df.loc[:, ~test_df.columns.str.contains("^Unnamed")]

# Separate input and output

X_train = train_df.drop("prognosis", axis=1)
y_train = train_df["prognosis"]

X_test = test_df.drop("prognosis", axis=1)
y_test = test_df["prognosis"]

print("Number of symptoms:", X_train.shape[1])
print("Number of diseases:", len(y_train.unique()))

# Train model

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

model.fit(X_train, y_train)

# Predictions

predictions = model.predict(X_test)

# Accuracy

accuracy = accuracy_score(y_test, predictions)

print("\nAccuracy:", round(accuracy * 100, 2), "%")

# Show some predictions

print("\nSample predictions:\n")

for i in range(10):

    print(
        "Actual:", y_test.iloc[i],
        "| Predicted:", predictions[i]
    )

# Test with custom symptoms

sample = [0] * len(X_train.columns)

# Example:
# fever = 1
# cough = 1

if "fever" in X_train.columns:
    sample[X_train.columns.get_loc("fever")] = 1

if "cough" in X_train.columns:
    sample[X_train.columns.get_loc("cough")] = 1

result = model.predict([sample])

print("\nPrediction for fever + cough:")

print(result[0])
```

```python
import pandas as pd

# Create an empty symptom vector

sample = [0] * len(X_train.columns)

# Example symptoms

selected_symptoms = [
    "fever",
    "cough"
]

# Set selected symptoms to 1

for symptom in selected_symptoms:

    if symptom in X_train.columns:

        index = X_train.columns.get_loc(symptom)

        sample[index] = 1

# Convert to DataFrame (warning disappears)

sample_df = pd.DataFrame(
    [sample],
    columns=X_train.columns
)

# Get probabilities

probabilities = model.predict_proba(sample_df)[0]

# Create disease-probability pairs

results = list(zip(model.classes_, probabilities))

# Sort in descending order

results = sorted(
    results,
    key=lambda x: x[1],
    reverse=True
)

print("Top 3 predictions:\n")

for disease, probability in results[:3]:

    print(
        f"{disease}: {probability * 100:.2f}%"
    )
```

```python
import re

# Get all symptoms from the dataset

all_symptoms = list(X_train.columns)

# User input

text = "I have chest pain and cough"

# Convert to lowercase

text = text.lower()

# Extract symptoms

detected_symptoms = []

for symptom in all_symptoms:

    symptom_text = symptom.replace("_", " ")

    pattern = r"\b" + re.escape(symptom_text) + r"\b"

    if re.search(pattern, text):

        detected_symptoms.append(symptom)

print("Detected symptoms:")

print(detected_symptoms)

# Create symptom vector

vector = [0] * len(all_symptoms)

for symptom in detected_symptoms:

    index = all_symptoms.index(symptom)

    vector[index] = 1

# Convert to DataFrame

sample_df = pd.DataFrame([vector], columns=all_symptoms)

# Predict top 3 diseases

probabilities = model.predict_proba(sample_df)[0]

results = list(zip(model.classes_, probabilities))

results = sorted(results, key=lambda x: x[1], reverse=True)

print("\nTop 3 predictions:\n")

for disease, probability in results[:3]:

    print(f"{disease}: {probability * 100:.2f}%")
```

```python
import pandas as pd

symptom_dict = {
    "high_fever": ["fever", "high fever", "temperature", "body hot"],
    "headache": ["headache", "head pain", "head hurts"],
    "stomach_pain": [
        "stomach pain",
        "stomach ache",
        "stomach hurts",
        "belly pain"
    ],
    "vomiting": ["vomiting", "vomit", "throwing up", "throw up"],
    "cough": ["cough", "coughing"],
    "chest_pain": ["chest pain", "pain in chest"],
    "breathlessness": [
        "breathlessness",
        "can't breathe",
        "shortness of breath"
    ]
}


def predict_from_text(text):

    text = text.lower()

    detected = []

    for symptom, phrases in symptom_dict.items():

        for phrase in phrases:

            if phrase in text:

                detected.append(symptom)
                break

    print("Detected symptoms:", detected)

    vector = [0] * len(X_train.columns)

    for symptom in detected:

        if symptom in X_train.columns:

            idx = X_train.columns.get_loc(symptom)

            vector[idx] = 1

    sample_df = pd.DataFrame([vector], columns=X_train.columns)

    probabilities = model.predict_proba(sample_df)[0]

    results = sorted(
        zip(model.classes_, probabilities),
        key=lambda x: x[1],
        reverse=True
    )

    print("\nTop 3 predictions:\n")

    for disease, prob in results[:3]:

        print(f"{disease}: {prob * 100:.2f}%")


# Test

predict_from_text(
    "I am throwing up and my stomach hurts"
)
```
